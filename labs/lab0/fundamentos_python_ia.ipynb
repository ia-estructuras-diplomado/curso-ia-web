## 0. Sintaxis esencial de Python

## 1. Paquetes, módulos e imports

- **Módulo:** un archivo `.py` (ej. `math.py` de la librería estándar).

### Formas de importar (debes reconocerlas en repos de ML)

```python
import pandas as pd          # paquete con alias corto
from pathlib import Path     # solo un nombre del módulo
import matplotlib.pyplot as plt
```

`import x as y` con alias corto (`pd`, `np`, `plt`) es la convención universal en ciencia de datos. `from modulo import nombre` trae solo ese nombre al espacio de nombres actual y evita colisiones cuando solo necesitas una función puntual (por ejemplo `from math import sqrt` vs. `import math` + `math.sqrt(...)`).

## 2. Estructuras de datos (listas y diccionarios)

Listas → secuencias (vectores/tensores más adelante). Diccionarios → JSON de APIs y agentes.

La convención de indexar desde 0 viene de C y se alinea con cómo NumPy/tensores indexan sus ejes. Una columna importada de un CSV es, conceptualmente, una lista ordenada de valores. Los diccionarios son intercambiables entre servicios como JSON — su estructura autodocumentada es justo lo que usarás para armar prompts y definir *tools* de agentes. Diferencia clave entre acceder con `.get()` y con `[]`: `.get()` devuelve `None` (o un valor por defecto) si la clave no existe, mientras que `[]` lanza `KeyError` — en código de producción casi siempre prefieres `.get()`.

## 3. Transformar datos (list comprehension)

Sintaxis compacta: `[expr for x in lista if condición]` — combina en una línea una expresión, una variable de iteración, un iterable y un filtro opcional. Es la forma "pythónica" de transformar listas, y la verás constantemente en código de limpieza de datos. Cuando la lógica es larga, tiene efectos secundarios (guardar archivos) o necesitas depurar paso a paso, un bucle `for` normal sigue siendo más claro — la comprehension es para transformaciones simples y legibles en una línea.


````python
precios = [100, 250, 45, 800, 120]
print('Originales:', precios)
````

````python
UMBRAL_DESCUENTO = 100
precios_con_descuento = [
    precio * 0.9 if precio > UMBRAL_DESCUENTO else precio
    for precio in precios
]
print('Tras regla:', precios_con_descuento)

n_desc = sum(1 for p in precios if p > UMBRAL_DESCUENTO)
print(f"ℹ️ {n_desc} producto(s) con descuento.")
````

## 4. Funciones = herramientas para agentes

Sintaxis: `def nombre(param: tipo) -> tipo:` + docstring + `return`. Esto no es solo estilo: cuando un agente de IA decide qué *tool* invocar, lee justamente el nombre de la función, sus type hints y su docstring para elegir la correcta — una función bien tipada y documentada es, literalmente, una herramienta legible tanto para humanos como para un LLM. El docstring en triple comilla justo bajo `def` documenta el propósito para ambos lectores; los type hints (`edad: int`, `-> str`) son un contrato que habilita autocompletado y linters como `mypy`, buena práctica en código de producción.


````python
def predecir_riesgo(edad: int, ingresos: float) -> str:
    """Evalúa riesgo: Bajo / Medio / Alto según reglas fijas."""
    if ingresos > 50000 and edad > 25:
        return "Riesgo Bajo"
    if ingresos < 20000:
        return "Riesgo Alto"
    return "Riesgo Medio"
````

````python
INGRESOS_PARA_PROBAR = 60_000
resultado_agente = predecir_riesgo(perfil_usuario["edad"], INGRESOS_PARA_PROBAR)
print('Tool →', resultado_agente)
````

## 5. Pandas — tablas para Machine Learning

Importado como `pd`. `DataFrame` = tabla; `Series` = columna. Un `DataFrame` se construye directamente desde un diccionario de listas: cada clave se vuelve una columna. En un problema de clasificación como este, `Compro` sería la variable objetivo (*y*) y `Edad`/`Salario` las *features* (*X*) — el `ID` normalmente se excluye del modelo. `df.describe()` resume estadísticas por columna y es la primera forma de detectar escalas muy distintas o valores anómalos. Filtrar con `df[condición]` crea una máscara booleana por fila y devuelve el subconjunto donde es `True`.


````python
data = {
    'ID': [1, 2, 3, 4],
    'Edad': [25, 30, 35, 40],
    'Salario': [50000, 60000, 75000, 90000],
    'Compro': [0, 0, 1, 1],
}
df = pd.DataFrame(data)
display(df)
display(df.describe())
````

````python
UMBRAL_SALARIO = 55_000
filtro = df[df['Salario'] > UMBRAL_SALARIO]
cuantas_compraron = int(filtro['Compro'].sum())
display(filtro)
print('Compraron:', cuantas_compraron)
````

## 6. Validación visual

**¿Por qué un ingeniero civil debería mirar un gráfico antes de confiar en una métrica como precisión?** Porque los datos pueden estar mal etiquetados, tener outliers o pocos registros — un gráfico revela patrones que un solo número resume mal. Este principio ("la IA propone, el ingeniero valida visualmente") es el que vas a usar en todos los labs siguientes.


````python
fig, ax = plt.subplots(figsize=(8, 4))
colores = df['Compro'].map({0: '#e74c3c', 1: '#27ae60'})
ax.bar([f'ID {i}' for i in df['ID']], df['Salario'] / 1000, color=colores)
ax.set_ylabel('Salario (miles)')
ax.set_title('Salario por cliente — verde = Compro (target)')
from matplotlib.patches import Patch
ax.legend(handles=[
    Patch(facecolor='#27ae60', label='Compró'),
    Patch(facecolor='#e74c3c', label='No compró'),
])
plt.tight_layout()
plt.show()
print("Compara el patrón visual con el filtro numérico de la sección 5: ¿coinciden?")
````
